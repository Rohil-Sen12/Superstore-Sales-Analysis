```python
top_city = toolkit.sort_values(by="Sales", ascending=False).head(5)
top_city
```

```python
fig = px.line(top_city,x="City",y="Profit",title="Profit Trend by City",template="plotly_dark")
fig.show()
```

```python
Top_5_customer_name = toolkit.sort_values(by="Sales",ascending=False).head(5)
Top_5_customer_name
```

```python
fig = px.pie(
    toolkit,
    names="Region",
    values="Sales",
    title="Sales Distribution by Region",
    template="plotly_white",
    color="Sales"
)
fig.update_traces(
hole=0.55, textinfo="percent+label",
 marker=dict(line=dict(color='black', width=2)),
 hovertemplate="<b>%{label}</b><br>Sales: %{value}<br>Share: %{percent}"
)
fig.show()
```

```python
fig = px.bar(Top_5_customer_name,x="Customer Name",y="Sales",color="Sales",text="Sales",template="plotly_white",title="Top 5 Customers by Sales")
fig.show()
```

```python
fig = px.scatter(toolkit,x="Category",y="Sales",color="Sales",template="plotly_dark",title="Sales Distribution by Category")
fig.show()
```

```python
fig= px.sunburst(toolkit,path=["Segment"],values="Profit",color="Profit",template="plotly_dark",title="Profit Distribution by Segment")
fig.show()
```

```python
top_5_city_sale = toolkit.sort_values(by="Sales",ascending=False).head(5)
top_5_city_sale
fig = px.funnel(top_5_city_sale,x="State",y="Sales",color="Sales",title="Top 5 States by Sales")
fig.update_traces(marker_line_width=2)
fig.update_layout(title_x=0.1)
fig.show()
```
